```python
import turicreate
song_data = turicreate.SFrame('../../data/song_data.sframe/')
song_data.print_rows(num_rows=4)
```

# Counting unique users who listened to an artist

```python
# 'Kanye West'
print ( 'Kanye West\t:', len ( song_data[song_data['artist'] == 'Kanye West']['user_id'].unique() ) )
# 'Foo Fighters'
print ( 'Foo Fighters\t:', len ( song_data[song_data['artist'] == 'Foo Fighters']['user_id'].unique() ) )
# 'Taylor Swift'
print ( 'Taylor Swift\t:', len ( song_data[song_data['artist'] == 'Taylor Swift']['user_id'].unique() ) )
# 'Lady GaGa'
print ( 'Lady GaGa\t:', len ( song_data[song_data['artist'] == 'Lady GaGa']['user_id'].unique() ) )
```

# Most popular and least popular artist

```python
popular_artists = song_data.groupby(key_column_names='artist', operations={'total_count': turicreate.aggregate.SUM('listen_count')})
print ( 'Most Popular Artist\t:', popular_artists.sort(key_column_names='total_count', ascending=False)[0] )
print ( 'Least Popular Artist\t:', popular_artists.sort(key_column_names='total_count', ascending=True)[0] )
```
