```python
# initialization
import turicreate as tc
image_train = tc.SFrame('../../data/image_train_data/')
image_test = tc.SFrame('../../data/image_test_data/')
```

```python
image_train[0:4].explore()
```

# Train a nearest-neighbors model for retrieving images using deep features

```python
knn_model = tc.nearest_neighbors.create(image_train, features=['deep_features'], label='id')
```

# Use image retrieval model with deep features to find similar images

```python
cat = image_train[18:19]
cat['image'].explore()
```

```python
knn_model.query(cat)
```

```python
def get_images_from_ids(query_result):
    return image_train.filter_by(query_result['reference_label'],'id')
```

```python
cat_neighbors = get_images_from_ids(knn_model.query(cat))
```

```python
cat_neighbors['image'].explore()
```

```python
car = image_train[8:9]
car['image'].explore()
```

```python
get_images_from_ids(knn_model.query(car))['image'].explore()
```

# Just for funm let's create a lambda to find and show nearest neighbot images

```python
show_neighbors = lambda i: get_images_from_ids(knn_model.query(image_train[i:i+1]))['image'].explore()
```
