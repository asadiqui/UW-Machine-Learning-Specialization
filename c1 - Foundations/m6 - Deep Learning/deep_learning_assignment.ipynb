```python
# initialization
import turicreate as tc
image_train = tc.SFrame('../../data/image_train_data/')
image_test = tc.SFrame('../../data/image_test_data/')
```

```python
image_train[0:4].explore()
```

# Computing summary statistics of the data

```python
image_train['label'].summary() # same as tc.Sketch(image_train['label'])
```

# Creating category-specific image retrieval models

```python
# Spliting train data by label
cat_data = image_train[image_train['label'] == 'cat']
dog_data = image_train[image_train['label'] == 'dog']
automobile_data = image_train[image_train['label'] == 'automobile']
bird_data = image_train[image_train['label'] == 'bird']
```

```python
# training each model
cat_model = tc.nearest_neighbors.create(cat_data, features=['deep_features'], label='id')
dog_model = tc.nearest_neighbors.create(dog_data, features=['deep_features'], label='id')
automobile_model = tc.nearest_neighbors.create(automobile_data, features=['deep_features'], label='id')
bird_model = tc.nearest_neighbors.create(bird_data, features=['deep_features'], label='id')
```

```python
# find the nearest neghibors to the first image in test data
nearest_cat = cat_model.query(image_test[0:1])
nearest_dog = dog_model.query(image_test[0:1])
```

```python
# printing the results as ints
print ( 'nearest cat-labeled image to the 1st image in the test data:', nearest_cat['reference_label'][0] )
print ( 'nearest dog-labeled image to the 1st image in the test data:', nearest_dog['reference_label'][0] )

# showing the result images
cat_data[cat_data['id'] == nearest_cat['reference_label'][0]].explore()
dog_data[dog_data['id'] == nearest_dog['reference_label'][0]].explore()
```

# A simple example of nearest-neighbors classification

```python
# calculating the mean of the 5-k nearest neghbors to the test image for cat_model and dog_model
print ( 'mean distance between image_test[0:1] at its 5 nearest neighbors that were labeled `cat`:', nearest_cat['distance'].mean() )
print ( 'mean distance between image_test[0:1] at its 5 nearest neighbors that were labeled `dog`:', nearest_dog['distance'].mean() )
```

# Computing nearest neighbors accuracy using SFrame operations

```python
# Spliting test data by label
image_test_cat = image_test[image_test['label'] == 'cat']
image_test_dog = image_test[image_test['label'] == 'dog']
image_test_automobile = image_test[image_test['label'] == 'automobile']
image_test_bird = image_test[image_test['label'] == 'bird']
```

```python
# the closest neighbor to each member of the dog test data using each of the trained models
dog_cat_neighbors = cat_model.query(image_test_dog, k=1) # k=1 finds the first neighbor
dog_dog_neighbors = dog_model.query(image_test_dog, k=1)
dog_automobile_neighbors = automobile_model.query(image_test_dog, k=1)
dog_bird_neighbors = bird_model.query(image_test_dog, k=1)
```

```python
# create an SFrame called dog_distances with 4 columns
dog_distances = tc.SFrame({
    'dog-dog' : dog_dog_neighbors['distance'],
    'dog-cat' : dog_cat_neighbors['distance'],
    'dog-automobile' : dog_automobile_neighbors['distance'],
    'dog-bird' : dog_bird_neighbors['distance']
    })
```

```python
# exploring the first 20 rows
dog_distances[0:19].explore()
```

```python
# function to check if dod-dog distances are the smallest
def is_dog_correct(row):
    if min(row['dog-dog'],
           row['dog-cat'],
           row['dog-automobile'],
           row['dog-bird']) == row['dog-dog']:
        return 1
    else:
        return 0
```

```python
# applying the function to the SFrame
dog_distances['is_correct'] = dog_distances.apply(is_dog_correct)
```

```python
# Accuracy of predicting dog in the test data
dog_distances['is_correct'].sum() / len(dog_distances) # 67.8 %
```
